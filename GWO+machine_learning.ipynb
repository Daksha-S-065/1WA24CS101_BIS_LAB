```python
import random

# User input
n = int(input("Enter number of students: "))

hours = []
marks = []

for i in range(n):
    h = float(input("Enter study hours: "))
    m = float(input("Enter marks: "))
    hours.append(h)
    marks.append(m)

# Fitness function (MSE)
def fitness(w, b):
    error = 0
    for i in range(n):
        predicted = w * hours[i] + b
        error += (marks[i] - predicted) ** 2
    return error / n

# Initialize wolves
N = 10
MaxIter = 50

wolves = [[random.uniform(0, 10), random.uniform(0, 50)]
          for _ in range(N)]

# GWO
for t in range(MaxIter):
    wolves.sort(key=lambda x: fitness(x[0], x[1]))

    Alpha = wolves[0][:]
    Beta = wolves[1][:]
    Delta = wolves[2][:]

    a = 2 - 2 * t / MaxIter
    new_wolves = []

    for wolf in wolves:
        new = []

        for d in range(2):
            r1, r2 = random.random(), random.random()
            A1 = 2 * a * r1 - a
            C1 = 2 * r2
            X1 = Alpha[d] - A1 * abs(C1 * Alpha[d] - wolf[d])

            r1, r2 = random.random(), random.random()
            A2 = 2 * a * r1 - a
            C2 = 2 * r2
            X2 = Beta[d] - A2 * abs(C2 * Beta[d] - wolf[d])

            r1, r2 = random.random(), random.random()
            A3 = 2 * a * r1 - a
            C3 = 2 * r2
            X3 = Delta[d] - A3 * abs(C3 * Delta[d] - wolf[d])

            new.append((X1 + X2 + X3) / 3)

        new_wolves.append(new)

    wolves = new_wolves

# Best solution
wolves.sort(key=lambda x: fitness(x[0], x[1]))
best = wolves[0]

print("\nBest Weight:", round(best[0], 2))
print("Best Bias:", round(best[1], 2))
print("Minimum MSE:", round(fitness(best[0], best[1]), 2))

# Prediction
h = float(input("\nEnter study hours to predict marks: "))
predicted = best[0] * h + best[1]

print("Predicted Marks:", round(predicted, 2))
```
