```python
import random

# User input
n = int(input("Enter number of jobs: "))

jobs = []
times = []

for i in range(n):
    t = int(input(f"Enter processing time of J{i+1}: "))
    jobs.append(f"J{i+1}")
    times.append(t)

# Fitness function: Total completion time
def fitness(schedule):
    total = 0
    completion = 0

    for job in schedule:
        completion += times[jobs.index(job)]
        total += completion

    return total

# Parameters
N = 10
MaxIter = 50

# Initialize cells with random job schedules
cells = []

for i in range(N):
    schedule = jobs.copy()
    random.shuffle(schedule)
    cells.append(schedule)

# Parallel Cellular Algorithm
for iteration in range(MaxIter):

    old_cells = [c.copy() for c in cells]
    new_cells = []

    for i in range(N):

        # Identify neighboring cells
        left = old_cells[(i - 1) % N]
        right = old_cells[(i + 1) % N]

        candidates = [
            old_cells[i].copy(),
            left.copy(),
            right.copy()
        ]

        # Create a new neighboring solution by swapping jobs
        new_schedule = old_cells[i].copy()

        if n > 1:
            a, b = random.sample(range(n), 2)
            new_schedule[a], new_schedule[b] = (
                new_schedule[b], new_schedule[a]
            )

        candidates.append(new_schedule)

        # Select the best schedule
        best_schedule = min(candidates, key=fitness)

        new_cells.append(best_schedule.copy())

    # Update all cells simultaneously
    cells = new_cells

# Find the best schedule
best = min(cells, key=fitness)

# Output
print("\nBest Job Schedule:", " -> ".join(best))
print("Minimum Total Completion Time:", fitness(best))

# Display completion times
completion = 0

print("\nJob\tProcessing\tCompletion")

for job in best:
    t = times[jobs.index(job)]
    completion += t
    print(job, "\t", t, "\t\t", completion)
```
