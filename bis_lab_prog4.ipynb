```python

import random

# User input
jobs = list(map(int, input("Enter job processing times: ").split()))

n = 10
pa = 0.25
MaxIter = 50

# Fitness function
def fitness(solution):
    completion = 0
    total = 0

    for i in solution:
        completion += jobs[i]
        total += completion

    return total

# Initialize nests
nests = []

for i in range(n):
    nest = list(range(len(jobs)))
    random.shuffle(nest)
    nests.append(nest)

# Find best nest
best = min(nests, key=fitness)
best_fit = fitness(best)

# Main loop
for iteration in range(MaxIter):

    # Generate new solution
    for i in range(n):
        new = nests[i][:]

        a, b = random.sample(range(len(jobs)), 2)
        new[a], new[b] = new[b], new[a]

        j = random.randint(0, n - 1)

        if fitness(new) < fitness(nests[j]):
            nests[j] = new

    # Abandon nests
    for i in range(n):
        if random.random() < pa:
            new = list(range(len(jobs)))
            random.shuffle(new)

            if fitness(new) < fitness(nests[i]):
                nests[i] = new

    # Update best solution
    current = min(nests, key=fitness)

    if fitness(current) < best_fit:
        best = current[:]
        best_fit = fitness(best)

# Display result
print("\nBest Job Scheduling:")

completion = 0

for i in best:
    completion += jobs[i]
    print("Job", i + 1, "Processing Time:", jobs[i],
          "Completion Time:", completion)

print("Minimum Total Completion Time:", best_fit)
```
